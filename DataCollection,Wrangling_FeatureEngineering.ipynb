```python
#we learn about User-defined, lambda, and built-in functions in Python

#user-defined functions can encapsulate a block of code that can be reused multiple times in a program. They are defined using the 'def' keyword followed by the function name and parentheses.
#lambda functions are anonymous functions that can be defined in a single line using the 'lambda' keyword. They are often used for short, simple operations.
#built in functions are pre-defined functions that come with Python and can be used without any additional code. Examples include 'print()', 'len()', and 'sum()'.

#USER DEFINED FUNCTION EXAMPLE 
#def Function_name(parameters1, parameters2):
    #z=parameters1 * parameters2
    #return z
    #multiply(parameters1, parameters2)  #this is how we call the function

#LAMBA FUNCTIONS

#z=lambda parameters1, parameters2: parameters1 * parameters2
#multiply= z(parameters1, parameters2)  #this is how we call the function

def multiply(x, y):
    z= x * y
    return z
    multiply(x, y)
print(multiply(5, 40))  # Output: 50

z= lambda x, y: x * y
multiply= z(5, 40)

# so if i have a small, single expression without having to give them a name using def keyword then i better using lambda function.


#EXAMPLE

effective_interest_rate= lambda interest_rate,periods : ((1+(interest_rate/periods))**periods)-1
effective_rate= effective_interest_rate(0.06, 12)
print('The effective interest rate is={}'. format(effective_rate))

amount_0= 10000 #amount at year 0
print('The total interest rate earned by the end of year 1=${}'.format(effective_rate*amount_0))
```

```python
#PERFORM FEATURE ENGINEERING

#Feature engineering involves transforming features in a dataset with the goal of improving the performance of downstream ML models. it involves feature selection, scaling, and creating new features from existing one.
# to calculate percentage Change calculation we can use Pandas .pct_change() method calculates the percentage change between current and prior elements in a Pandas DataFrame

#SYNTAX, For Example
#apple_df['Revenue' + 'change from last quarter']= apple_df['Revenue'].pct_change()

apple_df.info()
```

```python
#convert the 'Report Date' column to Datetime. this step is important when we want to sort the Pandas DataFrame and to perform date time operations
apple_df['Report Date']= pd.to_datetime(apple_df['Report Date'], format= '%m/%d/%Y')
apple_df=apple_df.sort_values(by='Report Date').reset_index(drop=True)
apple_df
```

```python
#now we try to perform feature engineering by creating a new feature that can capture the percentage change in quarterly revenue
#pct_change() calculates the change is percentage from last quarter to current quarter
#these new features might provide some insight to the model on how company performance changed form last quarter to this quarter

apple_df['Revenue' + 'change from last quarter']= apple_df['Revenue'].pct_change()
apple_df
```

```python
#now we can repeat the same process using additional features and by performing feature engineering by creating new features that capture the change in the net income, operating income, etc

features_to_get_delta = ['Shares (Basic)',
                         'Shares (Diluted)',
                         'Operating Income (Loss)',
                         'Net Income',
                         'Total Liabilities & Equity']
#pct_change() using for syntax
for feature in features_to_get_delta:
    apple_df[feature + 'change from last quarter']= apple_df[feature].pct_change()
apple_df
```

```python
#PANDAS SHIFT OPERATION

#pandas -shift() method shifts the index by a desired number of periods

#Syntax for pandas shift operation

#apple_df['Next Quarter EPS']= apple_df['Earning Per Share, Basic'].shift(-1)


apple_df['Next Quarter EPS'] = apple_df['Earnings Per Share, Basic'].shift(-1)
apple_df
```

```python
#drop NUll elements in Next Quarter EPS (last row)

apple_df = apple_df.dropna(subset = ['Next Quarter EPS']).reset_index(drop = True)
apple_df
```

```python
# Drop rows containing Null or missing values (first row)
apple_df = apple_df.dropna().reset_index(drop = True)
apple_df
```

```python
# The target label for the machine learning regression model is the quarterly percentage change in EPS

# Create a function that calculates the change in next quarter's EPS
def create_regression_label(next_quarter, current_quarter):
    
    # If the current quarter EPS value is zero, we set the label to zero 
    # Calculating the % change would result in an infinite number
    if abs(current_quarter) == 0:
        return 0
    return (next_quarter - current_quarter) / abs(current_quarter)

apple_df['% Quarterly Change in EPS (Regression Label)'] =\
apple_df.apply(lambda row: create_regression_label(row['Next Quarter EPS'], row['Earnings Per Share, Basic']), axis = 1)

apple_df
```
